```python
def bin_search(f, y, left, right, eps):
    while right - left > eps:
        mid = (left + right) / 2

        if f(mid) < y:
            left = mid
        else:
            right = mid

    return (left + right) / 2


def f(x):
    return x**3


print(bin_search(f, 1.5, 0, 10, 1e-4))
```
