```python

with open("input.txt", "r") as file:
    numbers = list(map(float, file.readline().split()))
    operation = file.readline().strip()

result = numbers[0]

for number in numbers[1:]:
    if operation == "+":
        result += number
    elif operation == "-":
        result -= number
    elif operation == "*":
        result *= number
    elif operation == "/":
        result /= number

with open("output.txt", "w") as file:
    file.write(str(result))
```

```python

n = int(input())

is_prime = [True] * (n + 1)
is_prime[0] = False

if n >= 1:
    is_prime[1] = False

for i in range(2, int(n ** 0.5) + 1):
    if is_prime[i]:
        for j in range(i * i, n + 1, i):
            is_prime[j] = False

for i in range(2, n + 1):
    if is_prime[i]:
        print(i, end=" ")
```
