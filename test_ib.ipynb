```python
questions = [
    ("1. Нужно ли использовать сложный пароль?\n1. Да\n2. Нет","1"),
    ("2. Что такое фишинг?\n"
     "A)Вид интернет-мошенничества\n"
     "B)Покупки в онлайн-магазине\n"
     "C)Реклама в интернете\n",
     "A"
    ),
    ("3. Что такое аутентификация?\n"
         "A)Процесс подтверждения личности пользователя\n"
         "B)Проверка целостности данных\n"
         "C)Шифрование данных при передаче\n",
         "A"
    ),
    ("4. Что такое антивирус?\n"
         "A)Молодёжное движение\n"
         "B)Программа для обнаружения компьютерных вирусов\n"
         "C)Противовирусные таблетки\n",
         "B"
    ),
]
score = 0
for question, correct_answer in questions:
    attempts = 5
    print("\n"+ question)

    while attempts > 0:
        user_answer = input("Введите букву правильного ответ: ").strip().upper()

        if user_answer == correct_answer:
            print("Верно!")
            score +=1
            break
        else:
            attempts -= 1
            if attempts > 0:
                print("Неверно, попробуйте еще раз")
                print("Осталось попыток: ",attempts)
            else:
                print("Попытки закончились.")
                print("Правильный ответ: ",correct_answer)
print("\nВопрос 5. В каком году родился Жорес Алферов?")
print("Подсказка: он родился в периуде c 1925-1935")

attempts = 5

while attempts > 0:
    user_answer = input("Введите год рождения: ").strip()

    if user_answer == "1930":
        points = attempts
        print("Верно!")
        print("Вы получили баллов:", points)
        score += points
        break
    else:
        attempts -= 1

        if attempts > 0:
            print("Неверно, попробуй еще раз")
            print("Осталось попыток: ", attempts)
        else:
            print("Попытки закончились")
            print("Правильный ответ: 1930")

print("\nТест завершен !")
print("Набрано баллов: ", score, "из 9")

if score >= 8:
    print("Оценка 5!")
elif score >= 6:
    print("Оценка 4!")
elif score >= 4:
    print("Оценка 3")
else:
    print("Оценка 2")
```
